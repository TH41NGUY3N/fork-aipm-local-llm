```python
import requests
import json
import sys
from pathlib import Path

OLLAMA_URL = "http://localhost:11434/api/generate"

# This script helps you understand how applications "talk" to AI models.
# You will implement a function to send a message to your local Ollama instance.

# Function to send a prompt to the local model
def query_ollama(model_name, prompt_text, timeout=30):
    """
    Args:
        model_name (str): The name of the local Ollama model (e.g., 'llama3.2:3b').
        prompt_text (str): The question or instruction for the AI.
        timeout (int): Seconds to wait before giving up.
        
    Returns:
        dict: A dictionary containing either the 'text' response or an 'error' message.
    """
    # The JSON payload required by the Ollama API contract
    payload = {
        "model": model_name,
        "prompt": prompt_text,
        "stream": False
    }

    try:
        # Send the POST request to the local server
        response = requests.post(OLLAMA_URL, json=payload, timeout=timeout)
        response.raise_for_status() # Raises an error for bad HTTP status codes (like 404 or 500)
        
        # Parse the JSON response
        response_json = response.json()
        
        # Common keys for textual output in Ollama
        response_text = response_json.get("response", "")
        
        # Return a structured dictionary
        return {"text": response_text, "error": None}
        
    except Exception as e:
        # If anything goes wrong, return the error gracefully
        return {"text": None, "error": str(e)}

# Function to save the result to a file (for documentation)
def save_output_to_file(filename, content):
    """Saves the generated text to a local file using pathlib."""
    file_path = Path(filename)
    file_path.write_text(content, encoding="utf-8")
    return True

# Main execution block
if __name__ == "__main__":

    print("Initializing Local AI Connection...")

    # Define our inputs
    my_model = "llama3.2:3b"
    my_prompt = "Explain Agile methodology in one sentence."

    # Call our function
    result = query_ollama(my_model, my_prompt)

    # Check for errors before trying to print or save
    if result.get("error"):
        print(f"An error occurred: {result['error']}")
        sys.exit(1) # Exit the script with an error code

    # Extract the successful text
    response_text = result.get("text")
    print(f"\n🤖 AI Response:\n{response_text}\n")

    # Save it to a file
    saved = save_output_to_file("ollama_output.txt", response_text)
    if saved:
        print("✅ Result successfully saved to 'ollama_output.txt'")
```

Step-by-Step Breakdown (The "Understand" Part)

Here is exactly what we solved in each section of the template:

1. The `payload` Dictionary
When applications talk to each other over the internet (or locally), they need a shared format. Ollama expects a JSON object containing the `model`, the `prompt`, and a boolean for `stream`. We set `"stream": False` so the API waits until the entire answer is generated before sending it back to our script.

2. The `try/except` Block
In the real world, APIs fail. The server might be down, or you might forget to run `ollama serve`. We use `requests.post()` to send the payload. If it fails, the `except Exception as e: `block catches the error so the script doesn't just crash abruptly; it returns a polite error message instead.

3. Parsing the `response_text`
Ollama sends back a large dictionary of metadata (how many tokens it used, how long it took, etc.). We only care about the actual text. By using `response_json.get("response", "")`, we safely extract just the generated words.

4. The `save_output_to_file` Function
Your coaches included `from pathlib import Path` at the top. This is the modern, Pythonic way to handle files. We create a Path object and use `.write_text()` to dump the AI's response into a `.txt` file for safe keeping.

5. The `__main__` Block
This is the command center. We define our variables, run the `query_ollama` function, check to make sure the `"error"` key is empty, print the result, and finally call our save function.

```python
import requests
import json

def query_ollama(prompt, model="llama3.2:3b", temperature=0.7):
    """
    Sends a prompt to the local Ollama API and returns the generated text.
    Includes the 'temperature' parameter for Step 4 of the assignment.
    """
    # This is the local port Ollama always runs on
    url = "http://localhost:11434/api/generate"
    
    # The JSON payload required by the Ollama API
    payload = {
        "model": model,
        "prompt": prompt,
        "stream": False,  # Set to False so we get the whole response at once, not word-by-word
        "options": {
            "temperature": temperature  # Step 4: Parameter Tuning!
        }
    }
    
    try:
        # Send the POST request to the local server
        response = requests.post(url, json=payload)
        response.raise_for_status() # Check if the request was successful
        
        # Parse the JSON response and extract the actual text
        result = response.json()
        return result.get("response", "Error: No text generated.")
        
    except requests.exceptions.ConnectionError:
        return "Error: Could not connect to Ollama. Is 'ollama serve' running?"
    except Exception as e:
        return f"An unexpected error occurred: {e}"

# ==========================================
# Testing the Application (Run this script!)
# ==========================================
if __name__ == "__main__":
    print("🚀 Initializing connection to Local LLM...\n")
    
    # Task 3 Test: Standard Request
    print("--- STANDARD TEST ---")
    standard_prompt = "Why is documentation important for an AI Project Manager? Answer in one sentence."
    print(f"Prompt: {standard_prompt}")
    print(f"Output: {query_ollama(standard_prompt)}\n")
    
    # Task 4 Test: Deterministic (Temp 0.0)
    print("--- TEMP 0.0 (Deterministic / Analytical) ---")
    data_prompt = "What is 2 + 2? Answer with just the number."
    print(f"Output: {query_ollama(data_prompt, temperature=0.0)}\n")
    
    # Task 4 Test: Creative (Temp 0.9)
    print("--- TEMP 0.9 (Creative / Brainstorming) ---")
    creative_prompt = "Write a highly creative, dramatic haiku about a software bug."
    print(f"Output: {query_ollama(creative_prompt, temperature=0.9)}\n")
```
